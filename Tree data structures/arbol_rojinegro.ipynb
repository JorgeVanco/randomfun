```python
class Nodo:
    def __init__(self, clave) -> None:
        self.color = 1 # 1-rojo 0-negro
        self.clave = clave
        self.padre = None
        self.der = None
        self.izq = None
    def __repr__(self):
        return f"Nodo({self.clave}, {'rojo' if self.color == 1 else 'negro'})"

    def imprimir(self):
        return f"Nodo({self.clave}, {'rojo' if self.color == 1 else 'negro'})"

class ArbolRojinegro:
    def __init__(self):
        self.raiz = None

    def anadirNodo(self, nodo_anadir):
        if not self.raiz:
            nodo_anadir.color = 0
            self.raiz = nodo_anadir

        else:
            nodo = self.raiz
            anadido = False
            while not anadido:
                if nodo_anadir.clave >= nodo.clave:
                    if nodo.der:
                        nodo = nodo.der
                    else:
                        nodo.der = nodo_anadir
                        anadido = True
                else:
                    if nodo.izq:
                        nodo = nodo.izq
                    else:
                        nodo.izq = nodo_anadir
                        anadido = True
            
            nodo_anadir.padre = nodo

            self.nivelar(nodo_anadir)


    def nivelar(self, nodo):
        # self.mostrar()
        if nodo.padre and nodo.padre.color == 1 and nodo.color == 1:  # Si el color del padre es rojo, hay que nivelar
            print("Nodo:", nodo, "Padre:", nodo.padre, "Abuelo:", nodo.padre.padre)
            padre = nodo.padre
            abuelo = padre.padre

            if abuelo.der == padre:
                tio = abuelo.izq
            else:
                tio = abuelo.der

            if not tio or tio.color == 0: # El tío es negro (nodos hoja(null) son negros)
                if abuelo.der == padre and padre.izq == nodo:
                    self.rotar_derecha(padre)
                elif abuelo.izq == padre and padre.der == nodo:  # Se trata de caso 2
                    self.rotar_izquierda(padre)
                # Caso 3
                
                if abuelo.der == padre and padre.der == nodo:
                    sentido = "izquierda"
                else:
                    sentido = "derecha"

                self.caso_3(nodo, padre, abuelo, tio, sentido)

            else: # El tío es rojo
                self.caso_1(nodo, padre, abuelo, tio)
            
        

    def caso_1(self, nodo, padre, abuelo, tio):
        abuelo.color = 1
        padre.color = 0
        tio.color = 0
        if self.raiz.color == 1:
            self.raiz.color = 0
        self.nivelar(abuelo)

    def caso_3(self, nodo, padre, abuelo, tio, sentido):
        padre.color = 0
        abuelo.color = 1
        if self.raiz.color == 1:
            self.raiz.color = 0
        if sentido == "derecha":
            self.rotar_derecha(abuelo)
        else:
            self.rotar_izquierda(abuelo)

        self.nivelar(padre)

    def rotar_derecha(self, nodo):
        nodo_izq = nodo.izq
        if nodo_izq:
            nodo.izq = nodo_izq.der
            nodo_izq.der = nodo

            nodo_izq.padre = nodo.padre
            nodo.padre = nodo_izq

            if not nodo_izq.padre:
                self.raiz = nodo_izq
            else:
                if nodo_izq.padre.clave < nodo_izq.clave:
                    nodo_izq.padre.der = nodo_izq
                else:
                    nodo_izq.padre.izq = nodo_izq

    def rotar_izquierda(self,nodo):
        nodo_der = nodo.der
        if nodo_der:
            nodo.der = nodo_der.izq
            nodo_der.izq = nodo

            nodo_der.padre = nodo.padre
            nodo.padre = nodo_der

            if not nodo_der.padre:
                self.raiz = nodo_der
            else:
                if nodo_der.padre.clave < nodo_der.clave:
                    nodo_der.padre.der = nodo_der
                else:
                    nodo_der.padre.izq = nodo_der

    def _checkear_imprimir(self, nodos_nivel_anterior):
        # Comprueba que haya algún nodo existente en la lista, para seguir recorriendo las capas del árbol
        imprimir  = False
        i = 0
        longitud_nivel_anterior = len(nodos_nivel_anterior)
        while i < longitud_nivel_anterior and not imprimir:
            if nodos_nivel_anterior[i] != None:
                imprimir = True
            i += 1
        return imprimir
    
    def altura_arbol(self):
        nivel = 0
        nodos_nivel = [self.raiz]
        bajar = self.raiz != None
        while bajar:   # Recorre los nodos de una capa, guarda los hijos en una lista para recorrerlos y bajar de nivel
            nodos_nivel_actual = []
            for nodo in nodos_nivel:
                if nodo:  # Si hay nodo, se guardan los hijos
                    nodos_nivel_actual.append(nodo.izq)
                    nodos_nivel_actual.append(nodo.der)
            nodos_nivel = nodos_nivel_actual
            if len(nodos_nivel_actual) == 0:
                bajar = False
            else:
                nivel += 1
        return nivel

    def mostrar(self, nodos_nivel_anterior = []):
        cadena = ""
        # Nodo(clave: 3)   -> 14 letras + 2*2 espacios
        altura = self.altura_arbol()
        nodos_ultima_capa = 2**(altura)

        longitud_cadena = len("Nodo(clave: 2)") + 4

        
        if len(nodos_nivel_anterior) == 0:
            cadena =  self.raiz.imprimir().center((nodos_ultima_capa*longitud_cadena)// 2, "_").center(nodos_ultima_capa*longitud_cadena) + "\n"
            cadena += "|".center(len(cadena)// 2) * 2 + "\n"
            cadena += self.mostrar([self.raiz])
            print(cadena)
            return #cadena
        
        imprimir = self._checkear_imprimir(nodos_nivel_anterior)
        
        if imprimir:
            nodos_nivel_actual = []
            cadena_union_nodos = ""

            for nodo in nodos_nivel_anterior:
                cadena_par_nodos = ""
                longitud_cadena_nodo = (nodos_ultima_capa*longitud_cadena) // (2 * len(nodos_nivel_anterior))  # La longitud de cadena de la última capa entre los nodos de esta capa (el doble de la capa anterior)

                if nodo: # Se guardan los hijos y se imprimen
                    cadena_nodo = ""
                    if nodo.izq:
                        cadena_nodo += nodo.izq.imprimir()
                        nodos_nivel_actual.append(nodo.izq)
                        cadena_union_nodos += "|".center(longitud_cadena_nodo// 2) * 2
                    else:
                        cadena_nodo += "None"
                        nodos_nivel_actual.append(None)
                        cadena_union_nodos += " ".center(longitud_cadena_nodo// 2) * 2
                    cadena_par_nodos += cadena_nodo.center(longitud_cadena_nodo// 2, "_").center(longitud_cadena_nodo)
                    
                    cadena_nodo = ""
                    if nodo.der:
                        cadena_nodo += nodo.der.imprimir()
                        nodos_nivel_actual.append(nodo.der)
                        cadena_union_nodos += "|".center(longitud_cadena_nodo// 2) * 2
                    else:
                        cadena_nodo += "None"
                        nodos_nivel_actual.append(None)
                        cadena_union_nodos += " ".center(longitud_cadena_nodo// 2) * 2

                    cadena_par_nodos += cadena_nodo.center(longitud_cadena_nodo// 2, "_").center(longitud_cadena_nodo)
                else:  # Si no hay nodo, se deja un espacio y se guarda None en la lista
                    cadena_par_nodos += " ".center(longitud_cadena_nodo * 2)
                    cadena_union_nodos += " ".center(longitud_cadena_nodo * 2)
                    nodos_nivel_actual.append(None)
                    nodos_nivel_actual.append(None)
                cadena_par_nodos = cadena_par_nodos.center(longitud_cadena_nodo// 2, "_").center(len(cadena_par_nodos))# + "   "
                cadena += cadena_par_nodos
            
            cadena = cadena.center(2**(altura - 1)*longitud_cadena)
            cadena += "\n"
            cadena += cadena_union_nodos + "\n"

            cadena += self.mostrar(nodos_nivel_actual)  # Se llama recursivamente la función para añadir el resto de capas
        return cadena
```

```python
arbol = ArbolRojinegro()
nodo100 = Nodo(100)
arbol.anadirNodo(nodo100)
arbol.mostrar()
for i in [150, 40, 74]:
    arbol.anadirNodo(Nodo(i))
    arbol.mostrar()
```
