## Árbol binario

```python
class Nodo:
    def __init__(self, clave) -> None:
        self.padre = None
        self.der = None
        self.izq = None
        self.clave = clave

    def __repr__(self) -> str:
        return f"Nodo(clave = {self.clave})"
    
    def imprimir(self) -> str:
        return f"Nodo(clave: {self.clave})"

class Arbol:
    def __init__(self) -> None:
        self.raiz = None

    def anadir_nodo(self, nodo_anadir):
        if self.raiz == None:
            self.raiz = nodo_anadir
        else:
            nodo = self.raiz
            anadido = False
            while not anadido:
                if nodo.clave < nodo_anadir.clave:
                    if nodo.der:
                        nodo = nodo.der
                    else:
                        nodo.der = nodo_anadir
                        anadido = True
                else:
                    if nodo.izq:
                        nodo = nodo.izq
                    else:
                        nodo.izq = nodo_anadir
                        anadido = True
            nodo_anadir.padre = nodo
    
    def mostrar_arbol(self, nodo = None):
        if not nodo:
            nodo = self.raiz
        
        if nodo.izq:
            self.mostrar_arbol(nodo.izq)
        print(nodo.clave)
        if nodo.der:
            self.mostrar_arbol(nodo.der)
    
    def buscar_nodo(self,clave):
        nodo = self.raiz
        while nodo and nodo.clave != clave:
            if nodo.clave > clave:
                nodo = nodo.izq
            else:
                nodo = nodo.der
        
        return nodo
        
    def maximo(self, nodo=None):
        if nodo == None:
            nodo = self.raiz

        while nodo.der:
            nodo = nodo.der
        return nodo

    def borrar_nodo(self,clave):
        nodo_borrar = self.buscar_nodo(clave)
        if nodo_borrar:
            if nodo_borrar.der != None and nodo_borrar.izq != None:
                nodo_intercambio = self.maximo(nodo_borrar.izq)
                # if nodo_intercambio.padre != nodo_borrar:
                #     nodo_intercambio.padre.der = nodo_intercambio.izq
                # else:
                #     nodo_borrar.izq = nodo_intercambio.izq
                self.borrar_nodo(nodo_intercambio.clave)
                nodo_intercambio.der = nodo_borrar.der
                nodo_intercambio.izq = nodo_borrar.izq
            
            elif nodo_borrar.der != None:
                nodo_intercambio = nodo_borrar.der 
            elif nodo_borrar.izq != None:
                nodo_intercambio = nodo_borrar.izq
            else:
                nodo_intercambio = None
            
            if nodo_borrar.padre:
                if nodo_borrar.clave > nodo_borrar.padre.clave:
                    nodo_borrar.padre.der = nodo_intercambio
                else:
                    nodo_borrar.padre.izq = nodo_intercambio
            else:
                self.raiz = nodo_intercambio
            if nodo_intercambio:
                nodo_intercambio.padre = nodo_borrar.padre

        return nodo_borrar

    def _checkear_imprimir(self, nodos_nivel_anterior):
        # Comprueba que haya algún nodo existente en la lista, para seguir recorriendo las capas del árbol
        imprimir  = False
        i = 0
        longitud_nivel_anterior = len(nodos_nivel_anterior)
        while i < longitud_nivel_anterior and not imprimir:
            if nodos_nivel_anterior[i] != None:
                imprimir = True
            i += 1
        return imprimir
    
    def altura_arbol(self):
        nivel = 0
        nodos_nivel = [self.raiz]
        bajar = self.raiz != None
        while bajar:   # Recorre los nodos de una capa, guarda los hijos en una lista para recorrerlos y bajar de nivel
            nodos_nivel_actual = []
            for nodo in nodos_nivel:
                if nodo:  # Si hay nodo, se guardan los hijos
                    nodos_nivel_actual.append(nodo.izq)
                    nodos_nivel_actual.append(nodo.der)
            nodos_nivel = nodos_nivel_actual
            if len(nodos_nivel_actual) == 0:
                bajar = False
            else:
                nivel += 1
        return nivel

    def mostrar(self, nodos_nivel_anterior = []):
        cadena = ""
        # Nodo(clave: 3)   -> 14 letras + 2*2 espacios
        altura = self.altura_arbol()
        nodos_ultima_capa = 2**(altura)

        longitud_cadena = len("Nodo(clave: 2)") + 4

        
        if len(nodos_nivel_anterior) == 0:
            cadena =  self.raiz.imprimir().center((nodos_ultima_capa*longitud_cadena)// 2, "_").center(nodos_ultima_capa*longitud_cadena) + "\n"
            cadena += "|".center(len(cadena)// 2) * 2 + "\n"
            cadena += self.mostrar([self.raiz])
            print(cadena)
            return #cadena
        
        imprimir = self._checkear_imprimir(nodos_nivel_anterior)
        
        if imprimir:
            nodos_nivel_actual = []
            cadena_union_nodos = ""

            for nodo in nodos_nivel_anterior:
                cadena_par_nodos = ""
                longitud_cadena_nodo = (nodos_ultima_capa*longitud_cadena) // (2 * len(nodos_nivel_anterior))  # La longitud de cadena de la última capa entre los nodos de esta capa (el doble de la capa anterior)

                if nodo: # Se guardan los hijos y se imprimen
                    cadena_nodo = ""
                    if nodo.izq:
                        cadena_nodo += nodo.izq.imprimir()
                        nodos_nivel_actual.append(nodo.izq)
                        cadena_union_nodos += "|".center(longitud_cadena_nodo// 2) * 2
                    else:
                        cadena_nodo += "None"
                        nodos_nivel_actual.append(None)
                        cadena_union_nodos += " ".center(longitud_cadena_nodo// 2) * 2
                    cadena_par_nodos += cadena_nodo.center(longitud_cadena_nodo// 2, "_").center(longitud_cadena_nodo)
                    
                    cadena_nodo = ""
                    if nodo.der:
                        cadena_nodo += nodo.der.imprimir()
                        nodos_nivel_actual.append(nodo.der)
                        cadena_union_nodos += "|".center(longitud_cadena_nodo// 2) * 2
                    else:
                        cadena_nodo += "None"
                        nodos_nivel_actual.append(None)
                        cadena_union_nodos += " ".center(longitud_cadena_nodo// 2) * 2

                    cadena_par_nodos += cadena_nodo.center(longitud_cadena_nodo// 2, "_").center(longitud_cadena_nodo)
                else:  # Si no hay nodo, se deja un espacio y se guarda None en la lista
                    cadena_par_nodos += " ".center(longitud_cadena_nodo * 2)
                    cadena_union_nodos += " ".center(longitud_cadena_nodo * 2)
                    nodos_nivel_actual.append(None)
                    nodos_nivel_actual.append(None)
                cadena_par_nodos = cadena_par_nodos.center(longitud_cadena_nodo// 2, "_").center(len(cadena_par_nodos))# + "   "
                cadena += cadena_par_nodos
            
            cadena = cadena.center(2**(altura - 1)*longitud_cadena)
            cadena += "\n"
            cadena += cadena_union_nodos + "\n"

            cadena += self.mostrar(nodos_nivel_actual)  # Se llama recursivamente la función para añadir el resto de capas
        return cadena
    
    def rutas(self, nodo = None, lista = None, path = ""):
        if nodo == None:
            nodo = self.raiz

        if lista == None:
            lista = []
        
        if nodo:
            if not nodo.der and not nodo.izq:
                lista.append(path + f"{nodo.clave}")
            else:
                path += f"{nodo.clave} -> "
                if nodo.izq:
                    self.rutas(nodo.izq, lista, path)
                if nodo.der:
                    self.rutas(nodo.der, lista, path)
        return lista
    
    def rotar_der(self, pivote = None):
        if pivote == None:
            pivote = self.raiz
        else:
            pivote = self.buscar_nodo(pivote)

        if pivote and pivote.izq:
            hijo_izq = pivote.izq
            rama_cambiar = hijo_izq.der
            hijo_izq.der = pivote
            pivote.izq = rama_cambiar
            hijo_izq.padre = pivote.padre

            if not pivote.padre:
                self.raiz = hijo_izq
            else:
                if pivote.clave > pivote.padre.clave:
                    pivote.padre.der = hijo_izq
                else:
                    pivote.padre.izq = hijo_izq

            pivote.padre = hijo_izq
    
    def rotar_izq(self, pivote:int = None):
        if pivote == None:
            pivote = self.raiz
        else:
            pivote = self.buscar_nodo(pivote)

        if pivote and pivote.der:
            hijo_der = pivote.der
            rama_cambiar = hijo_der.izq
            hijo_der.izq = pivote
            pivote.der = rama_cambiar
            hijo_der.padre = pivote.padre

            if not pivote.padre:
                self.raiz = hijo_der
            else:
                if pivote.clave > pivote.padre.clave:
                    pivote.padre.der = hijo_der
                else:
                    pivote.padre.izq = hijo_der

            pivote.padre = hijo_der

    def orden_central(self):
        pila = []
        imprimiendo = True
        nodo = self.raiz
        while imprimiendo:
            if nodo:
                pila.append(nodo)
                nodo = nodo.izq
            elif pila:
                nodo = pila.pop()
                print(nodo.clave)
                nodo = nodo.der
            else:
                imprimiendo = False

    def preorden(self):
        pila = []
        imprimiendo = True
        nodo = self.raiz
        while imprimiendo:
            if nodo:
                pila.append(nodo)
                nodo = nodo.izq
            elif pila:
                nodo = pila.pop()
                print(nodo.padre.clave)
                print(nodo.clave)
                pila.append(nodo.der)
            else:
                imprimiendo = False

    def preorden_recursivo(self, nodo = None):
        if nodo == None:
            nodo = self.raiz
        
        print(nodo.clave)
        if nodo.izq:
            self.preorden_recursivo(nodo.izq)
        if nodo.der:
            self.preorden_recursivo(nodo.der)

    def minimo(self, nodo= None):
        if nodo == None:
            nodo = self.raiz

        while nodo and nodo.izq:
            nodo = nodo.izq

        return nodo

    def sucesor(self, nodo):
        if nodo.der:
            return self.minimo(nodo.der)
        elif nodo.padre:
            padre = nodo.padre
            while padre and padre.clave < nodo.clave:
                padre = padre.padre
            return padre
        
    def postorden(self):
        if self.raiz is None:
            return
        
        postorden_lista = []
        nodo = self.raiz
        pila = []
        imprimir = True

        while imprimir:
            
            while nodo:
                if nodo.der:
                    pila.append(nodo.der)
                pila.append(nodo)
                nodo = nodo.izq
            
            nodo = pila.pop()
    
            if pila and nodo.der and pila[-1] == nodo.der:
                pila.pop()
                pila.append(nodo)
                nodo = nodo.der
            else:
                postorden_lista.append(nodo.clave)
                nodo = None
    
            if not pila:
                imprimir = False

        return postorden_lista

```
